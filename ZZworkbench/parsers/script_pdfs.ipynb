> extract pdfs to characters

```python
from pathlib import Path 

current_dir: Path = Path.cwd().resolve()
print(f"{current_dir = }")

root_dir: Path = current_dir.parents[1]
print(f"{root_dir = }")

pdfs_sheet_root: Path = root_dir / "inputs" / "pdfs_sheet"
pdfs_sheet_lst: list = list(pdfs_sheet_root.glob("*.pdf"))
print([x.name for x in pdfs_sheet_lst])
```

```python
from dotenv import load_dotenv
import os 
from pydantic import SecretStr
dotenv_path: Path = root_dir / ".env"
assert dotenv_path.is_file()
load_dotenv(dotenv_path=dotenv_path)
```

```python

mineru_api_key = SecretStr(os.getenv("MINERU_API_KEY"))
assert mineru_api_key is not None

# parse pdf-sheet  to character
from langchain_mineru import MinerULoader 
# from langchain_core.document_loaders import   xx  只有基类、抽象类 

muloader = MinerULoader(source=[str(pdfs_sheet_lst[0])])  # 默认 mode="flash"， 不需要token
docs = muloader.load()
print(type(docs))
print(len(docs))

for doc in docs:
    print(doc.page_content[:10])
    print(doc.metadata)
```

```python
# 更加推荐的方式

mineru_loader = MinerULoader(
    source=str(pdfs_sheet_lst[3]),
    mode="precision",
    token=mineru_api_key,  # pages="1-5",
    language="ch",
    split_pages=True,
    ocr=True,
    formula=True,
    table=True,
    timeout=3000,
)
   
mparsed_docs = mineru_loader.load() 
print(f"{type(mparsed_docs) = }")
print(f"{len(mparsed_docs) = }")
for doc in mparsed_docs:
    print()
    print(doc.metadata.get("page"))
    print(doc.page_content[:500])
```

```python
# langchain-paddleocr  
from langchain_paddleocr import PaddleOCRVLLoader 


paddleocr_api_key = SecretStr(os.getenv("PADDLEOCR_API_KEY"))
assert paddleocr_api_key is not None
pploader = PaddleOCRVLLoader(
    file_path=str(pdfs_sheet_lst[3]),
    model="PaddleOCR-VL-1.6",    # PaddleOCR-VL-1.6   【PaddleOCR-VL】  不支持：PP-OCRv6  PP-StructureV3，：官方web演示/使用是都支持的
    access_token=paddleocr_api_key
    
)
pp_docs = pploader.load()

print(f"Document 数量: {len(docs)}")

for doc in pp_docs:
    print(doc.metadata)
    print(doc.page_content[:1000])
```
