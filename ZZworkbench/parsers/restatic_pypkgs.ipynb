```python
from pathlib import Path 


# demo pdf file 
pdf_sheet_dir_src: Path = Path.cwd().resolve().parents[1] / "inputs" / "pdfs_sheet"
pdf_sheet_name: str = "三虎输变电工程三级进度计划土建部分.pdf"
pdf_sheet_path: Path = pdf_sheet_dir_src / pdf_sheet_name
assert pdf_sheet_path.is_file()
```

```python
import pdf_inspector 

# target dir 
pdf_sheet_dir_tgt: Path = Path.cwd().resolve().parents[1] / "outputs" / "parse_pdf_sheet" / "pdf-inspector"
pdf_sheet_dir_tgt.mkdir(parents=True, exist_ok=True)


result = pdf_inspector.process_pdf(str(pdf_sheet_path))
print(result.pdf_type)   # "text_based", "scanned", "image_based", "mixed"
print(result.markdown)   # Markdown string or None

# # Selective OCR; clean text PDFs do not load the external OCR runtime.
# ocr = pdf_inspector.process_pdf_with_ocr("document.pdf")
# print(ocr.pages_routed_to_ocr)
```
