一、 LangChain 核心组件与生态分层
LangChain 的架构采用模块化分层设计，上层编排依赖底层提供的标准化接口与异构集成能力：

LangChain Core & Components（核心与基础组件）：定义了统一的抽象基类与基础模块，包括 Chat models（聊天模型）、Agents（智能体编排）、Messages（消息结构）、Tools（工具绑定）、Short-term-memory（短期记忆与状态）、Event streaming / Streaming（流式传输）和 Structured output（结构化输出）。
LangChain Integrations（集成层）：提供与第三方大模型提供商（如 OpenAI, Anthropic）、向量数据库、嵌入模型、文档加载器（Document loaders）、文本分割器（Text Splitters）、沙箱环境（Sandboxes）以及检查点存储后端（Checkpoints / Stores）的标准化连接适配器。
```python
# 示例：通过核心模块初始化模型并绑定结构化输出与工具
```

二、 编排、短期记忆与持久化架构
LangChain 与 LangGraph 的组合通过严格的状态与记忆管理机制，解决了大模型在复杂多步任务中的“失忆”和“上下文窗口超限”问题：

Short-term memory（短期记忆）：基于对话线程（Thread）管理当前会话的消息历史与状态快照。
Checkpoints（检查点）：将图执行过程中的每一步状态持久化存储到关系型数据库（如 PostgreSQL）中，支持中断恢复（Human-in-the-loop）、时间旅行（Time-travel debugging）和断点续传。
Stores（长期/跨会话存储）：提供跨线程的用户档案、偏好设置或长期记忆存储，支持基于键值或向量检索的异步持久化。



三、 高级架构模式与多智能体（Advanced Usage）
在高级用例中，LangChain 提供了企业级应用所需的复杂架构模式：

Multi-agent（多智能体系统）：采用 Coordinator-Worker（协调者-工作者）或分层图架构，主智能体负责任务规划与拆解，将子任务分发给运行在隔离环境中的专业子智能体（如 Deep Agents 模式）。
Human-in-the-loop（人工介入）：在关键决策点（如高风险API调用、敏感数据修改）通过图中断机制（Interrupts）挂起执行，等待人工审批后再继续。
Model Context Protocol (MCP)：支持标准化连接外部数据源、工具和安全沙箱（如连接 JIRA、Slack、Confluence MCP 服务器）。
Guardrails & Retrieval（安全护栏与检索）：在输入/输出端加入安全过滤、PII脱敏与 RAG 检索增强（Retrievers, Vector stores, Embedding models）。



四、 生产交付与前端集成（Frontend & Production）
为了将 AI 智能体平稳推向生产环境，架构定义了清晰前后端分离与运维管道：

Frontend Integration（前端交互）：采用统一的后端 createAgent() 配合前端 useStream() 钩子，通过流式协议实时将状态、工具调用、子智能体快照渲染到 Web 或移动端界面。
Deployment & Observability（部署与可观测性）：通过 LangGraph Cloud / LangGraph CLI 或自托管方案（AWS/GCP Terraform 模块）进行容器化部署，结合 LangSmith 进行全链路分布式追踪（Tracing）、评估（Evaluation）与监控。

> LangGraph 的顶层架构设计逻辑核心在于将复杂的 AI 智能体交互建模为“具有状态图（State Graph）与循环控制能力的图计算引擎”，其设计灵感源自 Pregel 和 Apache Beam。
相比于传统 DAG（有向无环图）工作流框架，LangGraph 允许节点之间产生循环（Loops），这使其非常适合处理需要反复推理、工具调用、自我纠错和多智能体协作的复杂任务。

一、 核心架构基石：StateGraph 与 State Schema
LangGraph 的核心抽象是 StateGraph，整个应用围绕一个共享的**状态对象（State Schema）**运转：

State（状态定义）：通常由 TypedDict 或 Pydantic 模型定义。图中的每个节点都可以读取全局状态，并返回状态的更新（通过 Reducer 函数对历史消息或数据进行合并）。
Nodes（节点）：代表具体的计算单元（如 LLM 调用、工具执行、数据解析）。节点本质上是接收当前状态并输出状态更新的 Python/JS 函数。
Edges（边）：定义控制流。分为常规边（确定性跳转）和条件边 Conditional Edges（由 LLM 的输出或路由函数动态决定下一步走向）。
```python
from langgraph.graph import StateGraph, START, END
from typing_extensions import TypedDict

# 1. 定义状态
class State(TypedDict):
    messages: list

# 2. 初始化图构建器
builder = StateGraph(State)

# 3. 添加节点
def call_model(state: State):
    return {"messages": ["AI response"]}

builder.add_node("agent", call_model)
builder.add_edge(START, "agent")
builder.add_edge("agent", END)

# 4. 编译图
graph = builder.compile()
```


二、 持久化与状态管理（Checkpointers）
LangGraph 将“计算”与“状态存储”彻底解耦，通过 Checkpointer 机制实现了强大的企业级特性：

自动持久化：每次图节点执行完毕后，状态会自动写入检查点后端（如 PostgreSQL 或内存）。
Human-in-the-loop（人工介入与中断）：通过在关键节点配置 interrupt_before 或 interrupt_after，图会在执行到指定节点时自动挂起，等待用户审查或修改状态后安全恢复。
Time Travel（时间旅行）：由于每个检查点都带有唯一的 Thread ID 和版本号，开发者可以随时回滚到历史任意一步重新执行或调试。



三、 多智能体与子图隔离（Subgraphs & Multi-Agent）
随着应用复杂度的提升，LangGraph 的顶层架构原生支持模块化组合与多智能体协作：

子图隔离（Subgraphs）：复杂的业务逻辑可以拆分为多个独立的 StateGraph（例如规划子图、执行子图），然后作为节点嵌入到父图中。父子图之间通过明确的 State 映射进行通信，避免状态污染。
多智能体编排：支持主智能体（Coordinator）通过图路由将任务委派给专职的子智能体（Worker），每个子智能体拥有各自专用的工具集和系统提示词。



四、 生产运行时与云端服务（LangGraph Platform / API Server）
在生产交付层面，LangGraph 不仅是一个 Python/JS 开发库，还拥有完整的服务端运行时架构（LangGraph Server）：

声明式配置（langgraph.json）：通过配置文件指定依赖项、图入口、环境变量以及检查点和长期存储（Store）策略。
标准化 API 服务：编译后的图可以通过 LangGraph API Server 自动暴露出标准的 REST/Streaming 接口，无缝对接前端 SDK (useStream) 和 LangSmith 监控平台。

> LangSmith 的顶层架构设计逻辑核心在于构建一个面向大模型应用全生命周期（Observability、Evaluation、Testing、Deployment）的“控制面与数据面分离（Control Plane & Data Plane）”以及“分布式追踪与遥测”架构.

一、 控制面与数据面分离（Control Plane & Data Plane Model）
在企业级部署（如 BYOC / 自托管架构）中，LangSmith 采用了严格的双层架构，确保敏感应用数据的安全与隐私：

Control Plane（控制面）：运行在 LangChain 官方云端（如 us-east-2），负责身份认证（Authentication）、组织与工作空间管理、账单计费以及静态前端资源的托管。它不存储任何敏感的业务数据。
Data Plane（数据面）：运行在用户的私有云环境（如客户的 AWS / GCP 账号内，通过 EKS、Istio、RDS、S3 构建），负责摄取、存储和查询所有敏感的生产数据（包括 Traces 追踪日志、提示词、评估数据集、实验结果和代理部署）。
二、 全链路分布式追踪与遥测（Tracing & Telemetry）
LangSmith 的核心能力是捕获和重构 AI 应用的执行链路，其底层架构采用标准化的追踪协议：

非侵入式收集：通过 SDK 提供的 Tracer（如 LangChainTracer），在底层自动拦截并记录大模型调用、工具执行、检索（Retrieval）和图节点状态转换的完整输入、输出、耗时与 Token 消耗。
异步遥测管道：SDK 将追踪数据通过异步非阻塞的方式批量上报至数据面后端，确保不影响主业务链路的推理延迟。
三、 动态运行与代理部署（LangSmith Deployment & Studio）
LangSmith 的架构不仅用于“观察”，还深度融合了代理的运行与可视化调试：

Agent Server Protocol：定义了统一的代理服务标准协议（Agent Server API），支持将本地用 LangGraph 开发的智能体一键打包部署到生产环境。
Kubernetes Operator 架构：在自托管或云托管环境中，通过 langsmith-operator 和 LangGraphPlatform CRD 动态管理每个智能体实例的 Kubernetes Deployment 与服务生命周期。
LangSmith Studio：提供开箱即用的可视化 IDE，直接连接 Agent Server API，支持实时可视化调试图执行流、打断（Interrupt）与时间旅行。






> Deep Agents 和 Managed Deep Agents 的顶层架构设计逻辑核心在于构建一个“生产级、具备文件系统与子智能体派发能力的智能体 harness（代理运行骨架）”，它建立在 LangChain 核心组件与 LangGraph 运行时之上，专门解决复杂真实任务中的上下文膨胀、工具过载与长期规划难题。

一、 Deep Agents 的核心架构逻辑
Deep Agents 不是一个独立的新框架，而是 LangChain 生态中的高级代理 harness。它的顶层架构由四大支柱构成：

Execution Environment（执行环境与虚拟文件系统）： 提供标准化的文件系统后端（如 StateBackend、FilesystemBackend、CompositeBackend），允许智能体通过读写文件来存储中间状态、长文本或技能（Skills），从而摆脱纯内存上下文的限制。
Context Management（上下文与记忆管理）： 引入了渐进式工具披露（Progressive Tool Disclosure）、上下文卸载（Context Offloading）、摘要和提示词缓存机制，有效避免长对话导致的 Token 溢出与性能下降。
Delegation（任务规划与子智能体派发）： 采用 Coordinator-Worker（协调者-工作者）架构。主智能体（Coordinator）负责高层任务规划（Task Planning），并将具体的子任务（如代码审查、批量文档检索）并行派发（Fan-out and synthesize）给在独立隔离环境中运行的专业子智能体。
Steering（人工介入与控制）： 原生集成 LangGraph 的中断（Interrupts）与人工审批机制，确保高风险操作在执行前必须经过人类确认。



二、 Managed Deep Agents 的架构逻辑
当 Deep Agents 进入托管或生产交付环境时（如配合 LangSmith 平台），架构会扩展为 Managed 托管形态：

托管运行时（Managed Runtime）：底层的持久化存储、数据库、长期记忆（Store）以及沙箱执行环境由云端或企业自托管平台自动代管，开发者无需在代码中手动初始化复杂的数据库连接池。
前端流式与多智能体视图：通过 useStream() 钩子和子智能体选择器（Selector Helpers），前端不仅能接收主协调者的流式响应，还能独立订阅和渲染各个子智能体（Subagents）的专属运行状态与待办事项（Todos）。


1. `LangChain：底层构件库（The Foundation Layer）`
定位：底层标准与组件积木。
职责：它提供了整个 AI 开发所需的通用标准化接口（如 Chat models、Prompts、Messages、Tools、Document Loaders、Vector Stores）。它不强制规定你的应用必须怎么跑，而是提供最基础的“零件”。你可以用它来拼简单的 LLM 调用，也可以把它当作构建更高级框架的基石。
2. `LangGraph：编排运行时（The Orchestration & Runtime Layer）`
定位：Workflow / Chatflow 引擎（图计算与状态机）。
职责：当你的应用不再是单一的问答，而是需要多步循环（Loops）、条件跳转、持久化状态管理（Checkpointer）、人工审批（Human-in-the-loop）或复杂多智能体协作时，LangGraph 就登场了。它是底层构件之上的流程编排与状态运行时。
3. `Deep Agents：高阶智能体骨架/Harness（The Agent Harness Layer）`
定位：开箱即用的“全功能智能体系统”。
职责：虽然 LangGraph 给了你画图的能力，但要从零实现一个像高级程序员那样会用虚拟文件系统、会写代码解释器、会把任务拆解并动态派发给子智能体（Subagents）的复杂 Agent，依然需要写大量样板代码。Deep Agents 就是把这些复杂能力打包好的一套高阶 Agent Harness——它在 LangGraph 和 LangChain 的基础上，内置了文件系统后端、上下文管理、技能（Skills）和多智能体协调架构。
4. `Managed Deep Agents（托管 Deep Agents / 平台托管能力）：生产交付层（The Production / Managed Layer）`
定位：云端托管与服务端运行时。
职责：你的理解非常接近：它不是一个简单的 SDK，而是指在 LangSmith 平台或 LangGraph Cloud 上托管运行 Deep Agents 的能力。当你使用托管版时，平台会自动为你提供云端持久化 Store、数据库、自动扩缩容、安全沙箱以及与 LangSmith Studio/UI 的无缝对接，让开发者无需自行运维复杂的生产基础设施。

````python
import sys
from pathlib import Path

REPO_ROOT = next(
    path for path in (Path.cwd(), *Path.cwd().parents)
    if (path / 'knowledge' / 'project_progress' / 'texts').is_dir()
)
if str(REPO_ROOT) not in sys.path:
    sys.path.insert(0, str(REPO_ROOT))

from ZZworkbench.rag_langchain.text_retrieval import (
    DEFAULT_INDEX_ROOT,
    ChunkConfig,
    EmbeddingConfig,
    build_embeddings,
    chunk_statistics,
    evaluate_retriever,
    load_eval_cases,
    load_txt_documents,
    open_chroma,
    retrieve_hybrid,
    split_documents,
)
````

````python
chunk_config = ChunkConfig(chunk_size=872, chunk_overlap=160)
documents = load_txt_documents(version='v4')
chunks = split_documents(documents, chunk_config)

print(f'documents={len(documents)}, chunks={len(chunks)}')
print(chunk_statistics(chunks))
display(documents[0])
display(chunks[0])
````

````python
embedding_config = EmbeddingConfig()
embeddings = build_embeddings(embedding_config)
vector_store = open_chroma(DEFAULT_INDEX_ROOT, embeddings)

query = '珠海黄金输变电工程的主体结构封顶计划什么时候完成？'
hits = retrieve_hybrid(vector_store, query, k=4)
for hit in hits:
    print(hit.rank, round(hit.score, 6), hit.document.metadata['source'])
    print(hit.document.page_content, '\n')
````

````python
eval_path = REPO_ROOT / 'knowledge' / 'project_progress' / 'evals' / 'retrieval_v4.jsonl'
report = evaluate_retriever(vector_store, load_eval_cases(eval_path), k=4)
print({key: value for key, value in report.items() if key != 'rows'})
````
